```python
def imprimir_sequencia(n):
    for i in range(1, n + 1):
        linha = ''.join(str(i) for _ in range(i))
        print(linha)
n = int(input("digite um número inteiro n:"))
imprimir_sequencia(n)
```

```python
def positivo_negativo(n):
    if n > 0:
        return 'P'
    else:
        return 'N'
numero = int(input("digite um numero:"))
resultado = positivo_negativo(numero)
print(f"o resultado para {numero} é:{resultado}")
```
